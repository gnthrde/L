```python
import numpy as np
import matplotlib.pyplot as plt

# Define L and arealradius R

L = 1
def R(r, a):
    return a / (1 + L*(a**2 * r**2))

# range of values / grid / function z(x,y)
r = np.linspace(-3, 3, 300)
a = np.linspace(-3, 3, 300)
x, y = np.meshgrid(r, a)
z = R(x, y)

# clipping

z_cut = np.clip(z, -5, 5)

# 3d-plot
fig = plt.figure(figsize=(20, 8))
ax = fig.add_subplot(1,2,1, projection='3d')

surf = ax.plot_surface(x, y, z_cut, cmap='viridis', edgecolor='none')

# define fontsize fs / axis
fs=19
ax.set_xlabel('r' , fontsize=fs)
ax.set_ylabel('a' , fontsize=fs)
ax.set_zlabel('$ R(a,r) $' , fontsize=fs)
ax.set_title('$ L=1 $' , fontsize=fs)



ax2 = fig.add_subplot(1, 2, 2)
contour = ax2.contourf(x, y, z_cut, levels=30, cmap='viridis')
ax2.set_xlabel('a', fontsize=fs)
ax2.set_ylabel('r', fontsize=fs)
fig.colorbar(contour, ax=ax2)

plt.tight_layout()
plt.show()
```
